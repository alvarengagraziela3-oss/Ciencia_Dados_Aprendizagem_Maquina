```python
import pandas as pd

# ==========================================
# 1. Series e DataFrame
# ==========================================

prioridades = pd.Series(
    ["Alta", "Média", "Baixa"],
    name="prioridade"
)

exemplo = pd.DataFrame({
    "id": ["CH001", "CH002"],
    "tempo": [3.5, 12.0]
})

print("SERIES:")
print(prioridades)

print("\nDATAFRAME DE EXEMPLO:")
print(exemplo)


# ==========================================
# 2. Carregamento e inspeção
# ==========================================

df = pd.read_csv("chamados_suporte_ti.csv")

print("\nPRIMEIROS 5 REGISTROS:")
print(df.head())

print("\nÚLTIMOS 3 REGISTROS:")
print(df.tail(3))

print("\nDIMENSÕES:")
print(df.shape)

print("\nCOLUNAS:")
print(df.columns.tolist())

print("\nINFORMAÇÕES:")
df.info()

print("\nESTATÍSTICAS DESCRITIVAS:")
print(df.describe())


# ==========================================
# 3. Seleção
# ==========================================

print("\nCOLUNA SETOR:")
print(df["setor"])

print("\nID E SETOR:")
print(df[["id_chamado", "setor"]])

print("\nLOC:")
print(df.loc[0:2, ["id_chamado", "setor"]])

print("\nILOC:")
print(df.iloc[0:3, 0:2])


# ==========================================
# 4. Filtros
# ==========================================

print("\nPRIORIDADE ALTA:")
print(df[df["prioridade"] == "Alta"])

print("\nPRIORIDADE ALTA E RESOLVIDO:")
print(
    df[
        (df["prioridade"] == "Alta") &
        (df["status"] == "Resolvido")
    ]
)

print("\nCATEGORIAS ACESSO OU REDE:")
print(
    df[
        df["categoria"].isin(["Acesso", "Rede"])
    ]
)


# ==========================================
# 5. Transformações
# ==========================================

df["data_abertura"] = pd.to_datetime(
    df["data_abertura"]
)

df["tempo_dias"] = (
    df["tempo_resolucao_horas"] / 24
)

df["satisfacao"] = df["satisfacao"].astype("Int64")

print("\nDATAFRAME APÓS TRANSFORMAÇÕES:")
print(df.head())


# ==========================================
# 6. Agregações
# ==========================================

print("\nQUANTIDADE POR STATUS:")
print(df["status"].value_counts())

print("\nTEMPO MÉDIO DE RESOLUÇÃO:")
print(df["tempo_resolucao_horas"].mean())

print("\nTEMPO MÉDIO POR CATEGORIA:")
print(
    df.groupby("categoria")[
        "tempo_resolucao_horas"
    ].mean().sort_values(ascending=False)
)
```
